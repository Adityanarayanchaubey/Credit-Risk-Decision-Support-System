## Loading dataset 1 (borrower's data)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = pd.read_csv("../data/raw/borrower_profiles.csv")
```

```python
df1
```

```python
df1.isna().sum() #no null values
```

```python
#data types of each column
df1.dtypes
```

```python
print(df1.state.unique())
print(df1.education_level.unique())
print(df1.employment_status.unique())
print(df1.home_ownership.unique())
```

```python
categorical_columns=[
    'state',
    'education_level',
    'employment_status',
    'home_ownership'
]

df1[categorical_columns]=df1[categorical_columns].astype('category')
```

```python
df1.dtypes
```

```python
print("================================BASIC STATS REPORT OF BORROWER'S DATA===============================")
print(f"1.rows: {df1.shape[0]}, cols:{df1.shape[1]}")
print("-"*100)
print("2.columns=",df1.columns)
print("-"*100)
print("3.missing values are:\n",df1.isna().sum())
print("-"*100)
print("4.data types of each columns are as follows:\n",df1.dtypes)
print("-"*100)
print("5.Preview of data are as follows:\n",df1.head())
print("-"*100)
print("Basic description of data:\n",df1.describe())
```
