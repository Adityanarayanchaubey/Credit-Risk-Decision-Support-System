```python
import pandas as pd
```

```python
df=pd.read_csv("../data/processed/cleaned_loan_data.csv")
```

```python
df
```

## Column 1: monthly income

```python
df["monthly_income"]=df["annual_income"]/12
df
```

## column 2: loan_income_ratio

```python
df["loan_income_ratio"]=df["loan_amount"]/df["annual_income"]
```

```python
df
```

## column 3: high_dti

```python
df["high_dti"]=(df["dti_ratio"]>30).astype(int)
```

```python
df
```

## column 4: income_per_dependent

```python
df["income_per_dependent"]=df["annual_income"]/(df["dependents"]+1)
```
