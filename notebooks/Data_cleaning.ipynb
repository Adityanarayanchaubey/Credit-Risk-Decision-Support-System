```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_borrower=pd.read_csv("../data/raw/borrower_profiles.csv")
df_loan=pd.read_csv("../data/raw/loan_applications.csv")
```

```python
df_borrower.head()
```

```python
df_loan.head()
```

```python
df=pd.merge(df_borrower,df_loan)
```

```python
df
```

## Removing unwanted column

```python
df.drop(columns=["borrower_id","loan_id","application_date","loan_status"],inplace=True)
```

## Encoding categorical columns

```python
df=pd.get_dummies(
    df,
    columns=["state","employment_status","home_ownership","loan_purpose"],
    dtype=int
    
)
```

```python
df
```

```python
df["education_level"].unique()
education_map={
    "High School":1,
    "Associate":2,
    "Bachelor":3,
    "Master":4,
    "Doctorate":5
}
df["education_level"]=df["education_level"].map(education_map)
    
```

```python
df
```

## checking mising value


```python
for col in df.columns:
    print(f"{col}:",df[col].isna().sum())
```

## Checking data types of each column

```python
for col in df.columns:
    print(f"{col}:",df[col].dtypes)
```
